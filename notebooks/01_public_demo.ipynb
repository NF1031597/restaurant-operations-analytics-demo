# Restaurant Staffing & Demand — Public Demo

**Synthetic data only.** No employer-owned data, identifying information, or private records appear in this notebook.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

path = Path('data/sample/synthetic_hourly.csv')
if not path.exists():
    path = Path('/content/synthetic_hourly.csv')
assert path.exists(), 'Upload synthetic_hourly.csv to Colab or run in repository root'
demo = pd.read_csv(path, parse_dates=['date'])
print('SYNTHETIC records:', len(demo))
display(demo.head())
```

## Compare demand with coverage
We use only **preliminary** entries and divide total example orders by total example staff-hours. This ratio is not a staffing requirement.


```python
valid = demo.loc[(demo['coverage_quality'] == 'preliminary') & (demo['kitchen_staff_hours'] > 0)].copy()
valid['weekday'] = valid['date'].dt.day_name()
summary = valid.groupby(['weekday', 'hour']).agg(orders=('positive_orders','sum'), staff_hours=('kitchen_staff_hours','sum')).reset_index()
summary['ratio'] = summary['orders'] / summary['staff_hours']
display(summary.head(15).round(2))
```

```python
pivot = summary.pivot(index='weekday',columns='hour',values='ratio').reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
fig,ax = plt.subplots(figsize=(12,5))
image = ax.imshow(pivot.to_numpy(),aspect='auto',cmap='YlOrBr')
ax.set_xticks(range(len(pivot.columns)),pivot.columns)
ax.set_yticks(range(len(pivot.index)),pivot.index)
ax.set_title('SYNTHETIC Example — Hourly Workload')
ax.set_xlabel('Hour of day')
fig.colorbar(image, ax=ax, label='Orders per kitchen staff-hour')
plt.tight_layout()
plt.show()
```
